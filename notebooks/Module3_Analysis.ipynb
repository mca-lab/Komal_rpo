```python
#Distribution of rentals
plt.figure(figsize=(8,5))
sns.histplot(day_df["total_rentals"], kde=True)
plt.title("Distribution of Daily Bike Rentals")
plt.xlabel("Total Rentals")
plt.show()
```

```python
# Rental by Season
plt.figure(figsize=(8,5))
sns.barplot(x='season', y='total_rentals', data=day_df)
plt.title("Average Bike Rentals by Season")
plt.xlabel("Season")
plt.ylabel("Total Rentals")
plt.show()
```

```python
# Temperature vs Rentals scatter plot
plt.figure(figsize=(8,5))
sns.scatterplot(x='temp', y='total_rentals', data=day_df)
plt.title("Bike Rentals vs Temperature")
plt.xlabel("Temperature (normalized)")
plt.ylabel("Total Rentals")
plt.show()
```

```python
# Corelation Heatmap
plt.figure(figsize=(10,6))

numeric_df = day_df.select_dtypes(include=['float64','int64'])

sns.heatmap(numeric_df.corr(), annot=False, cmap="coolwarm")
plt.title("Correlation Heatmap (numeric columns only)")
plt.show()
```

```python
# Boxplot Working vs Non working days
plt.figure(figsize=(8,5))
sns.boxplot(x='workingday', y='total_rentals', data=day_df)
plt.title("Rentals on Working vs Non Working Days")
plt.xlabel("Working Day (0 = No, 1 = Yes)")
plt.ylabel("Total Rentals")
plt.show()
```

```python
# Aggregate hourly data to daily sum
hour_daily = hour_df.groupby(['yr','mnth','weekday','season','holiday','workingday','weathersit']).agg({
    'temp':'mean',
    'atemp':'mean',
    'hum':'mean',
    'windspeed':'mean',
    'casual':'sum',
    'registered':'sum',
    'total_rentals':'sum'
}).reset_index()

# Merge with day_df on matching columns
merged_df = pd.merge(day_df, hour_daily, 
                     on=['yr','mnth','weekday','season','holiday','workingday','weathersit'], 
                     suffixes=('_day','_hour'))
```

```python
merged_df.head()
merged_df.shape
```

```python
# Compute Correlation
numeric_cols = merged_df.select_dtypes(include=['int64','float64'])
corr_matrix = numeric_cols.corr()

# Correlation heatmap
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,8))
sns.heatmap(corr_matrix, cmap="coolwarm", annot=True, fmt=".2f")
plt.title("Correlation Heatmap - Merged Dataset")
plt.show()
```

```python
#Analyse Correlation with Rental
# Correlation of all features with total_rentals_day
corr_with_rentals = corr_matrix['total_rentals_day'].sort_values(ascending=False)
print(corr_with_rentals)
```
